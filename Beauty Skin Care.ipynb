# Beauty and Skin care prediction system

## 2. Business Understanding

### 2.1 Business Problem
* Consumers have many beauty and cosmetic products to choose from, but selecting an 
appropriate product can be difficult because products differ in price, ratings, ingredients, skin-type suitability, usage frequency, and other characteristics.
* At the same time, beauty businesses need to understand which types of products are popular, 
highly rated, and likely to attract customers.
* Therefore, there is a need for a data-driven system that can analyze product information and 
identify important patterns that can support better product selection and business decisio-making

### 2.2 Project Aim

* The main aim of this project is to analyze beauty and cosmetics product data and develop a 
data-driven product recommendation or prediction system that can help consumers 
identify suitable products while providing useful business insights.

### 2.3 Specific Objectives

The project will:
1.  Identify the most popular product categories and brands.
2.  Analyze the relationship between product prices, ratings, and number of reviews.
3.  Determine which products are associated with different skin types.
4. Investigate the relationship between usage frequency and product characteristics.
5.  Identify common ingredients used in highly rated products.
6. Analyze consumer preferences across different product categories.
7.  Develop a machine learning model for product recommendation or prediction.
8.  Evaluate the performance of the developed model.
9.  Provide recommendations that can support consumers and beauty businesses.

### 2.4 Key Business Questions

The project will attempt to answer questions such as:
1.  Which beauty product categories are most common?
2.  Which brands have the highest-rated products?
3. Does price influence product ratings?
4.  Which products receive the largest number of reviews?
5. Which ingredients are most frequently used?

## 3. Data Understanding

- The variables include both numerical and categorical information.
1. Numerical Variables
* Price_USD – price of the product in US dollars.

*  Number_of_Reviews – number of reviews associated with the product.
2. Categorical Variables
* Product_Name
* Brand
* Category
* Usage_Frequency
* Product_Size
* Skin_Type
*  Gender_Target
*  Packaging_Type
*  Main_Ingredient
* Country_of_Origin
3. Boolean Variable
* Cruelty_Free – indicates whether the product is cruelty-fre

## Data Visualizations.


```python
## Identify the Catergories  represented most in the dataset and visualize it using a bar chart
plt.figure(figsize=(10, 5))

df["Category"].value_counts().plot(
    kind="bar",
    edgecolor="black"
)

plt.title("Distribution of Cosmetic Product Categories")
plt.xlabel("Category")
plt.ylabel("Number of Products")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

### Findings
Serum is the most used in all of the products displayed while lip gloss is least used.

```python
#plotting the distribution of prices of cosmetic products using a histogram
plt.figure(figsize=(8, 5))
sns.histplot(df["Price_USD"].dropna(), bins=20, kde=True)
plt.title("Distribution of Cosmetic Product Prices")
plt.xlabel("Price (USD)")
plt.ylabel("Number of Products")
plt.show()
```

### Findings 
- Dataset contains cosmetic products across a broad range of prices,with a relatively even distribution across price intervals.
- Highly suggesting that the dataset represents different price segments rather than concentrating mainly on low-priced or high-priced products. 

```python
# Compare which product categories are more prevalent across different skin types using a count plot
plt.figure(figsize=(11, 6))

sns.countplot(
    data=df,
    x="Skin_Type",
    hue="Category"
)

plt.title("Cosmetic Categories Across Skin Types")
plt.xlabel("Skin Type")
plt.ylabel("Number of Products")
plt.xticks(rotation=30)
plt.legend(title="Category", bbox_to_anchor=(1.02, 1))
plt.tight_layout()
plt.show()
```
