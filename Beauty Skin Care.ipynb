## Data Visualizations.


```python
## Identify the Catergories  represented most in the dataset and visualize it using a bar chart
plt.figure(figsize=(10, 5))

df["Category"].value_counts().plot(
    kind="bar",
    edgecolor="black"
)

plt.title("Distribution of Cosmetic Product Categories")
plt.xlabel("Category")
plt.ylabel("Number of Products")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

### Findings
Serum is the most used in all of the products displayed while lip gloss is least used.

```python
#plotting the distribution of prices of cosmetic products using a histogram
plt.figure(figsize=(8, 5))
sns.histplot(df["Price_USD"].dropna(), bins=20, kde=True)
plt.title("Distribution of Cosmetic Product Prices")
plt.xlabel("Price (USD)")
plt.ylabel("Number of Products")
plt.show()
```

### Findings 
- Dataset contains cosmetic products across a broad range of prices,with a relatively even distribution across price intervals.
- Highly suggesting that the dataset represents different price segments rather than concentrating mainly on low-priced or high-priced products. 

```python
# Compare which product categories are more prevalent across different skin types using a count plot
plt.figure(figsize=(11, 6))

sns.countplot(
    data=df,
    x="Skin_Type",
    hue="Category"
)

plt.title("Cosmetic Categories Across Skin Types")
plt.xlabel("Skin Type")
plt.ylabel("Number of Products")
plt.xticks(rotation=30)
plt.legend(title="Category", bbox_to_anchor=(1.02, 1))
plt.tight_layout()
plt.show()
```

### Findings
* The count plot identifies which skin types have the highest and lowest numbers of products.

* A higher count indicates that more products in the dataset are associated with that skin type.

* A lower count may indicate fewer products are listed for that skin type.

* This analysis can help businesses identify potential gaps in their product offering

```python


# Box plot of product rating distribution
plt.figure(figsize=(8, 5))
sns.boxplot(y=df["Rating"])

plt.title("Distribution of Product Ratings")
plt.ylabel("Product Rating")
plt.show()
```

### Findings
- Boxplot shows that product rating range approximately 1.0 to 5.0 indicating varying levels of customer satisfaction.
- Median rating is approximately 3.0 hence show half of the ratings fall below this valueand half above it.
- The middle 50% of product ratings lie approximately between 2.0 and 4.1, showing considerable variation in customer opinions.
- No obvious outliers are visible in the box plot, suggesting there are no clearly isolated extreme rating according to this visualizations. 

```python
plt.figure(figsize=(12, 6))

sns.barplot(
    data=df,
    x="Category",
    y="Rating",
    hue="Skin_Type",
    errorbar=None
)

plt.title("Average Product Rating by Category and Skin Type")
plt.xlabel("Product Category")
plt.ylabel("Average Rating")
plt.xticks(rotation=45)
plt.legend(title="Skin Type", bbox_to_anchor=(1.05, 1),
           
           loc="upper left")

plt.tight_layout()
plt.show()
```

### Findings
- Average product ratings are relatively similar across product categories and skin types, with most ratings around 3.0. 
- Businesses can use these findings to identify lower-rated categories and investigate how to improve customer satisfaction for different skin type.

```python

# Calculate brand performance metrics
brand_performance = df.groupby("Brand").agg(
    Average_Price=("Price_USD", "mean"),
    Average_Rating=("Rating", "mean"),
    Total_Reviews=("Number_of_Reviews", "sum"),
    Number_of_Products=("Product_Name", "count")
).reset_index()

# Visualize brand performance
plt.figure(figsize=(12, 7))

sns.scatterplot(
    data=brand_performance,
    x="Average_Price",
    y="Average_Rating",
    size="Total_Reviews",
    hue="Average_Rating",
    sizes=(50, 500),
    palette="viridis",
    alpha=0.8
)

plt.title("Brand Performance: Customer Satisfaction vs Pricing")
plt.xlabel("Average Product Price (USD)")
plt.ylabel("Average Customer Rating")
plt.tight_layout()
plt.show()
```

How to interpret the graph:
- High rating, low price: Brands offering relatively affordable products with high average ratings.
- High rating, high price: Brands with higher-priced products and high average ratings.
- Low rating, high price: Brands that may need to review product quality, value, or customer expectations.
- Low rating, low price: Affordable brands whose products may have opportunities for improvement.
- Larger circles: Brands with more reviews across their products, based on their total review count.

```python

# Select numerical columns
numeric_cols = ["Price_USD", "Rating", "Number_of_Reviews"]

# Calculate correlations
correlation = df[numeric_cols].corr()

# Plot heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(correlation, annot=True, cmap="coolwarm",
            fmt=".2f", vmin=-1, vmax=1)

plt.title("Correlation Heatmap of Beauty Products")
plt.tight_layout()
plt.show()
```
